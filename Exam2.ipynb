```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model  import LinearRegression
from sklearn.ensemble      import RandomForestRegressor
from sklearn.metrics       import mean_absolute_error, r2_score
```

```python
model_df = df.copy()
# One-hot encoding: turns text into numbers
# drop_first=True removes one column to avoid redundancy
model_df['Profit Margin (%)'] = model_df['Profit Margin (%)'].fillna(model_df['Profit Margin (%)'].mean())
model_df = pd.get_dummies(model_df, columns=['Category', 'Region'], drop_first=True, dtype = int)

print('Columns after encoding:')
print(model_df.columns.tolist())
print('\nShape:', model_df.shape)
```

```python
model_df['Total Sales'] = model_df['Units Sold'] * model_df['Unit Price']
```

```python

y = model_df['Total Sales']


X = model_df.drop(columns=['Total Sales'])

X = X.select_dtypes(include='number')

print('Input columns (X):', X.columns.tolist())
print('Target column (y): Total Sales')
```

```python
# Split data: 80% for training, 20% for testing
# random_state=42 just means the split is the same every time you run it
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.2,       # 20% goes to test
    random_state=42
)

print(f'Training rows : {len(X_train)}')
print(f'Testing rows  : {len(X_test)}')
```

```python
lr_model = LinearRegression()

# .fit() = train it on the training data
lr_model.fit(X_train, y_train)

# .predict() = use the trained model to guess Total Sales on the test rows
lr_preds = lr_model.predict(X_test)

# MAE = Mean Absolute Error → on average, how many $ off is the model?
lr_mae = mean_absolute_error(y_test, lr_preds)

# R² score → how much of the variation does the model explain?
# 1.0 = perfect, 0.0 = no better than guessing the average
lr_r2 = r2_score(y_test, lr_preds)

print(f'Linear Regression  MAE: ${lr_mae:,.0f}   R²: {lr_r2:.3f}')
```

```python
# n_estimators = how many trees to build (more = more accurate but slower)
# random_state = same as before, keeps results consistent
rf_model = RandomForestRegressor(n_estimators=100, random_state=42)

# Train it
rf_model.fit(X_train, y_train)

# Predict on test rows
rf_preds = rf_model.predict(X_test)

rf_mae = mean_absolute_error(y_test, rf_preds)
rf_r2  = r2_score(y_test, rf_preds)

print(f'Random Forest   MAE: ${rf_mae:,.0f}   R²: {rf_r2:.3f}')
```

```python
results = pd.DataFrame({
    'Model': ['Linear Regression', 'Random Forest'],
    'MAE ($)': [lr_mae, rf_mae],
    'R² Score': [lr_r2, rf_r2]
})
```

```python
results
```

```python
# feature_importances_ gives a score for each column
# higher score = that column was more useful for predictions
importances = pd.Series(
    rf_model.feature_importances_,
    index=X_train.columns
)

# Sort from most to least important
importances = importances.sort_values(ascending=True)

# Horizontal bar chart
plt.figure(figsize=(8, 5))
plt.barh(importances.index, importances.values, color='steelblue')
plt.title('Which columns helped the model most?')
plt.xlabel('Importance score')
plt.tight_layout()
plt.show()
```

```python
from sklearn.tree import DecisionTreeRegressor
# Train a Decision Tree with NO depth limit first
# max_depth=None means it grows until every leaf has 1 sample → classic overfit
dt_model = DecisionTreeRegressor(max_depth=None, random_state=42)
dt_model.fit(X_train, y_train)

dt_train_preds = dt_model.predict(X_train)   # predict on TRAINING data
dt_test_preds  = dt_model.predict(X_test)    # predict on TEST data

train_r2 = r2_score(y_train, dt_train_preds)
test_r2  = r2_score(y_test,  dt_test_preds)

print(f'Unlimited depth → Train R²: {train_r2:.3f}  |  Test R²: {test_r2:.3f}')
print()
print('If Train R² > Test R², the model is OVERFITTING.')
```
